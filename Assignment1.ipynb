**5** : **GPA Calculator**

```python
n = int(input("Enter number of courses:"))
gp_sum = 0
sum_credits = 0
for i in range(n):
    course_name = input(f"Enter name of course {i+1}: ")
    credits = int(input(f"Enter credits for {course_name}: "))
    grade = float(input(f"Enter grade for {course_name} (0-10 scale): "))
    gp_sum += grade * credits
    sum_credits += credits

print("GPA of the student is:", gp_sum / sum_credits)
```

**6** : **Functions and Iteration**

```python
def Factorial(n):
    ans  = 1
    for i in range(1,n+1):
        ans *= i
    return ans

def Fibonacci(n):
    a, b = 0, 1
    fib_sequence = []
    for _ in range(n):
        fib_sequence.append(a)
        a, b = b, a + b
    return fib_sequence

def is_Prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

print("Factorial of 5:", Factorial(5))
print("Fibonacci sequence of 10:", Fibonacci(10))
print("Is 29 a prime number?", is_Prime(29))
```

**8** : **Exception Handling**

```python
def divide_numbers(a, b):
    try:
        result = a / b
        return result
    except ZeroDivisionError:
        return "Error: Division by zero is not allowed."

print(divide_numbers(10, 2))  
print(divide_numbers(10, 0))  
```

**9** : **List Comprehension**

```python
squares = [x**2 for x in range(1, 21)]
even_numbers = [x for x in range(1, 51) if x % 2 == 0]
divisible_by_3_and_5 = [x for x in range(1, 101) if x % 3 == 0 and x % 5 == 0]
print("Squares of numbers from 1 to 20:", squares)
print("Even numbers from 1 to 50:", even_numbers)
print("Numbers divisible by both 3 and 5 from 1 to 100:", divisible_by_3_and_5)
```

**10** : **Mini Programming Problem**

```python
students = {}

while True:
    print("\n----- STUDENT RECORD MENU -----")
    print("1. Add Student")
    print("2. Update Marks")
    print("3. Search Student")
    print("4. Delete Student")
    print("5. Display All Students")
    print("6. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        name = input("Enter student name: ")
        marks = int(input("Enter marks: "))
        students[name] = marks
        print("Student added successfully!")

    elif choice == 2:
        name = input("Enter student name: ")
        if name in students:
            marks = int(input("Enter new marks: "))
            students[name] = marks
            print("Marks updated successfully!")
        else:
            print("Student not found!")

    elif choice == 3:
        name = input("Enter student name: ")
        if name in students:
            print("Marks =", students[name])
        else:
            print("Student not found!")

    elif choice == 4:
        name = input("Enter student name: ")
        if name in students:
            del students[name]
            print("Student deleted successfully!")
        else:
            print("Student not found!")

    elif choice == 5:
        if len(students) == 0:
            print("No student records available.")
        else:
            print("\nStudent Records:")
            for name, marks in students.items():
                print(name, ":", marks)

    elif choice == 6:
        print("Program Ended.")
        break

    else:
        print("Invalid Choice! Try Again.")
```
